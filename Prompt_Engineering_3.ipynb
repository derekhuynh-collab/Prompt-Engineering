**Initial prompt used:** Summarize the customer feedback below for a support manager in one or two clear sentences. Use only facts in the passage.

“Customers liked the fast checkout and friendly staff. Several said delivery arrived late, and two reported damaged packaging. They asked for clearer tracking updates.”

**Self-critique prompt used:** Compare your first summary with the passage. Check accuracy, clarity, usefulness for a support manager, missing details, unsupported claims, plain-sentence formatting, and the one-to-two-sentence limit. Explain what needs improvement. Then write a revised summary that keeps the important praise, complaints, and customer request without adding facts.

**First summary:** Customers liked the service but had delivery problems.

**Improvement:** The first summary was vague. The revised summary includes fast checkout, friendly staff, late delivery, damaged packaging, and the request for clearer tracking.

Paste into a code cell and run:

```python
feedback = (
    "Customers liked the fast checkout and friendly staff. Several said "
    "delivery arrived late, and two reported damaged packaging. They "
    "asked for clearer tracking updates."
)

first_summary = "Customers liked the service but had delivery problems."

required_details = {
    "fast checkout": "checkout",
    "friendly staff": "staff",
    "late delivery": "late",
    "damaged packaging": "damaged",
    "tracking updates": "tracking"
}

missing = [
    detail for detail, keyword in required_details.items()
    if keyword not in first_summary.lower()
]

critique = (
    "The first summary is too vague. It leaves out: "
    + ", ".join(missing)
    + ". It does not make unsupported claims, but it needs more specific details."
)

improved_summary = (
    "Customers praised fast checkout and friendly staff. Some reported "
    "late delivery, two reported damaged packaging, and they wanted "
    "clearer tracking updates."
)

print("Original feedback:", feedback)
print("\nFirst summary:", first_summary)
print("\nSelf-critique:", critique)
print("\nImproved summary:", improved_summary)
```
