
**Prompt used:** Act as a customer support assistant. A customer says, “My package arrived damaged. Can I get a replacement?” Follow these steps:

1. Identify the issue and the customer's request. Return a category and requested action.
2. Use that result to write a friendly response of no more than two sentences. Ask for an order number and a photo. Do not guarantee a replacement.
3. Review the response for clarity and unsupported promises, then give the final reply.

**How the chain works:** The customer's message goes into Step 1. Step 1's category goes into Step 2. Step 2's draft goes into Step 3.

**Improvement made:** An earlier response promised a replacement immediately. The final response asks for information so the request can be reviewed first.



```python
customer_message = "My package arrived damaged. Can I get a replacement?"

def identify_issue(message):
    if "damaged" in message.lower():
        return {"category": "damaged order", "request": "replacement"}
    return {"category": "general question", "request": "more information"}

def draft_response(issue):
    if issue["category"] == "damaged order":
        return (
            "I'm sorry your package arrived damaged. Please send your "
            "order number and a photo so we can review a replacement."
        )
    return "Could you share more details so we can help?"

def review_response(reply):
    if "guaranteed" in reply.lower():
        return "Please contact support so we can review your request."
    return reply

issue = identify_issue(customer_message)
draft = draft_response(issue)
final_reply = review_response(draft)

print("Customer:", customer_message)
print("Step 1 - Issue:", issue)
print("Step 2 - Draft:", draft)
print("Step 3 - Final reply:", final_reply)
```
