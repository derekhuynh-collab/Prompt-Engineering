**Full prompt used:** Create Python 3 code that calculates the average of [5, 4, 3, 5, 4]. Use only built-in Python functions and return None for an empty list. Show each stage separately:

- PLAN: Briefly describe how the function will work.
- ACT: Write and run the code using the ratings and an empty list.
- OBSERVE: Show the actual results or any errors.
- FIX: If the code fails or a result is wrong, correct it and run it again.
- FINAL: State the verified average to one decimal place.

**Improvement made:** A first version divided by the length of the list without checking whether it was empty. The revised function checks for an empty list before dividing.

```python
print("PLAN: Calculate the average and handle an empty list.")

def average_rating(ratings):
    if not ratings:
        return None
    return sum(ratings) / len(ratings)

ratings = [5, 4, 3, 5, 4]

print("ACT: Run the function with", ratings, "and [].")
result = average_rating(ratings)
empty_result = average_rating([])

print("OBSERVE: Average =", result)
print("OBSERVE: Empty list =", empty_result)

if result != 4.2 or empty_result is not None:
    print("FIX: The results need correction.")
else:
    print("FIX: No further correction needed.")

print(f"FINAL: The average rating is {result:.1f} out of 5.")
```
